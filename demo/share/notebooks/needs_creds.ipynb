```python
# Pide credenciales por consola: el launcher las responde solo.
# Sirve para verificar dos cosas a la vez:
#   1. que SID/PASSWORD se inyecten bien
#   2. que el redactor tape el password aunque el notebook lo imprima
import getpass
import os

print("Preparando la subida a DTC...")
```

```python
usuario = input("Usuario de red: ")
clave = getpass.getpass("Contrasena de red: ")
print(f"Autenticado como {usuario}")
```

```python
# A proposito: se intenta filtrar el password de las tres formas posibles.
print("Desde el entorno  ->", os.environ.get("PASSWORD"))
print("Desde getpass()   ->", clave)
print("Interpolado       -> user=%s&pass=%s" % (usuario, clave))
print()
print("Si arriba se ven puntos y no la clave, el redactor esta haciendo su trabajo.")
```
